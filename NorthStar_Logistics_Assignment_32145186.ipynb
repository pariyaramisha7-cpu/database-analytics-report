## 1. Python-Based Operational Analytics
This section analyzes the delivery performance, hub efficiency, and incident patterns using Python libraries.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Loading Data from your assignment tables
# Table 2: Delivery Status
status_data = {'Status': ['On-Time', 'Delayed', 'Failed'], 'Count': [1850, 450, 150]}
# Table 3: Hub Performance
hub_data = {'Hub': ['London', 'Manchester', 'Birmingham', 'Bristol', 'Leeds'],
            'Deliveries': [1200, 950, 800, 600, 750]}
# Table 8: Incident Types
incident_data = {'Type': ['Vehicle Fault', 'Traffic', 'Weather', 'Accident'], 'Count': [45, 30, 15, 10]}

df_status = pd.DataFrame(status_data)
df_hub = pd.DataFrame(hub_data)
df_incident = pd.DataFrame(incident_data)

# 2. Creating Graphs
plt.figure(figsize=(16, 5))

# Plot 1: Delivery Status (Pie Chart)
plt.subplot(1, 3, 1)
plt.pie(df_status['Count'], labels=df_status['Status'], autopct='%1.1f%%', colors=['#2ecc71', '#f1c40f', '#e74c3c'])
plt.title('Figure 1: Delivery Status Distribution')

# Plot 2: Hub Performance (Bar Chart)
plt.subplot(1, 3, 2)
sns.barplot(x='Hub', y='Deliveries', data=df_hub, palette='viridis')
plt.title('Figure 2: Hub-Level Performance')

# Plot 3: Incident Distribution (Bar Chart)
plt.subplot(1, 3, 3)
plt.bar(df_incident['Type'], df_incident['Count'], color='salmon')
plt.xticks(rotation=45)
plt.title('Figure 6: Incident Distribution')

plt.tight_layout()
plt.show()

print("Operational Analytics Completed.")
```
