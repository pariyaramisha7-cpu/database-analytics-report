```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Delivery Status Data (from Table 2)
delivery_data = {'Status': ['On-Time', 'Delayed', 'Failed'], 'Count': [616, 202, 132]}
df_delivery = pd.DataFrame(delivery_data)

# 2. Hub-Level Performance Data (from Table 3)
hub_perf = {
    'Hub': ['North Exchange', 'South Link', 'East Dock', 'West Gate', 'Central Core', 'Airport Hub', 'Riverside Hub', 'Midtown Relay'],
    'On_Time_Rate': [68.38, 66.04, 71.43, 65.35, 58.26, 59.62, 66.09, 62.50]
}
df_hub = pd.DataFrame(hub_perf)

# 3. Visualizations
plt.figure(figsize=(16, 6))

# Figure 1: Overall Delivery Reliability
plt.subplot(1, 2, 1)
plt.pie(df_delivery['Count'], labels=df_delivery['Status'], autopct='%1.2f%%', colors=['#27ae60', '#f1c40f', '#e74c3c'], startangle=140)
plt.title('Figure 1: Delivery Status Distribution')

# Figure 2: Hub Performance Comparison
plt.subplot(1, 2, 2)
sns.barplot(x='On_Time_Rate', y='Hub', data=df_hub.sort_values('On_Time_Rate', ascending=False), palette='viridis')
plt.axvline(60, color='red', linestyle='--', label='60% Efficiency Floor')
plt.title('Figure 2: Hub-Level On-Time Performance (%)')
plt.xlabel('Success Rate (%)')
plt.legend()

plt.tight_layout()
plt.show()
```
